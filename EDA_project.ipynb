```python
make_encode={'alfa-romero': 1 , 'audi':2, 'bmw':3 , 'chevrolet':4, 'dodge':5, 'honda':6, 'isuzu':7, 'jaguar':8,
 'mazda':9, 'mercedes-benz': 10,'mercury':11, 'mitsubishi':12, 'nissan':13, 'peugot':14, 'plymouth':15,
 'porsche':16, 'renault':17, 'saab':18, 'subaru':19, 'toyota':20, 'volkswagen':21, 'volvo':22}

fuel_type_encode={'diesel':1,'gas':2}
aspiration_encode={'std':1,'turbo':2}
body_style_encode={'convertible':1,'hatchback':2,'sedan':3,'wagon':4,'hardtop':5}
drive_wheels_encode={'rwd':1,'fwd':2,'4wd':3}
engine_location_encode={'front':1,'rear':2}
engine_type_encode={'dohc':1,'ohcv':2,'ohcf':3,'l':4,'rotor':5,'ohc':6,'dohcv':7}
num_of_cylinders_encode={'four':4,'six':6,'five':5,'eight':8,'two':2,'twelve':12,'three':3}
fuel_system_encode={'mpfi':1,'2bbl':2,'mfi':3,'1bbl':4,'spdi':5,'4bbl':6,'idi':7,'spfi':8}

df['make']=df['make'].map(make_encode)
df['fuel-type']=df['fuel-type'].map(fuel_type_encode)
df['aspiration']=df['aspiration'].map(aspiration_encode)
df['body-style']=df['body-style'].map(body_style_encode)
df['drive-wheels']=df['drive-wheels'].map(drive_wheels_encode)
df['engine-location']=df['engine-location'].map(engine_location_encode)
df['engine-type']=df['engine-type'].map(engine_type_encode)
df['num-of-cylinders']=df['num-of-cylinders'].map(num_of_cylinders_encode)
df['fuel-system']=df['fuel-system'].map(fuel_system_encode)
```
